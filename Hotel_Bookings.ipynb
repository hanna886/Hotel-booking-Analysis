## UNIVARIATE ANALYSIS

```python
data
```

### Hotel Type

```python
plt.figure(figsize=(6,4))
sns.countplot(data=data,x='hotel',palette='BuGn_r')
plt.title("Hotel Type")
plt.xlabel("Type of hotel")
plt.ylabel("Count")
plt.show()
```

City Hotel type was most frequent used type by customers

### MONTH OF ARRIVALS

```python
sns.histplot(data['arrival_date_month'],color='skyblue')
plt.xticks(rotation=90)
plt.title("Month of arrivals")
plt.show()
```

It is the month of Augest that has highest arrivals

### Reservation Status

```python
data["reservation_status"].value_counts().plot(kind="pie",colors=["skyblue","yellow","pink"],autopct="%1.1f%%",startangle=90)
plt.title("Reservation status")
plt.show()
```

Cancelled status has a rate of 26.9% while checkout has 72%

## BIVARIATE ANALYSIS

### Time Lead in Hotels

```python
plt.figure(figsize=(6, 6))
plt.bar(data["hotel"],data["lead_time"], color='pink')
plt.title('Time lead in Hotels')
plt.xlabel('Hotel')
plt.ylabel('Time lead')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

It is in Resort Hotels that has the most lead time while compared to City Hotel.

### Lead Time on Different Months

```python
july_month=data[data["arrival_date_month"]=="July"]
```

```python
hd=july_month.head(30)
```

```python
sns.countplot(x=hd['lead_time'],palette='pastel')
plt.xticks(rotation=90)
plt.title("lead time on july") 
plt.show()
```

```python
april_month=data[data["arrival_date_month"]=="April"]
```

```python
hd1=april_month.head(30)
```

```python
sns.countplot(x=hd1['lead_time'],palette='pastel')
plt.xticks(rotation=90)
plt.title("lead time on april") 
plt.show()
```

July has a highest lead time of 37 and April has 171

```python
data["arrival_date_year"]
```

### Rate of adults in arrival dates

```python
plt.figure(figsize=(12, 5))
sns.countplot(data=data, x="arrival_date_year", hue='adults', order=data['arrival_date_year'].value_counts().index, palette='Set2')
plt.title('Count of adult')
plt.xlabel('Arraival Year')
plt.ylabel('count')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### Boxplot of stayes in weekends in every year

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=data, x='arrival_date_year', y='stays_in_weekend_nights', palette='Set2')
plt.title('stays in wekends in every year')
plt.xlabel('Year')
plt.ylabel('stay in wwekend')
plt.tight_layout()
plt.show()
```

Median of stayes in weekend is almost same for every year
